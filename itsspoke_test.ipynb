```python
from pathlib import Path
import re


root = Path(
    "data/itspoke_uncertainty/PSLC-WOZ-allfiles"
)

log_path = (
    root
    / "borglogs"
    / "borg_pslc01-c1_58.log"
)

transcript_path = (
    root
    / "studenttranscripts-sos"
    / "pslc01-c1.txt"
)


# ============================================================
# TIME HELPERS
# ============================================================

def transcript_time_to_seconds(value):
    """
    Converts:
        0:00:48.682
    to:
        48.682
    """

    parts = value.strip().split(":")

    hours = int(parts[0])
    minutes = int(parts[1])
    seconds = float(parts[2])

    return (
        hours * 3600
        + minutes * 60
        + seconds
    )


def log_time_to_seconds(value):
    """
    BORG timestamps appear to be milliseconds.

    Example:
        0000055038 -> 55.038 seconds
    """

    return int(value) / 1000.0


# ============================================================
# LOAD STUDENT TRANSCRIPT
# ============================================================

student_turns = []

for line in transcript_path.read_text(
    encoding="utf-8",
    errors="replace",
).splitlines():

    parts = line.split(",", 2)

    if len(parts) != 3:
        continue

    start_raw, end_raw, text = parts

    student_turns.append({
        "start_sec":
            transcript_time_to_seconds(start_raw),

        "end_sec":
            transcript_time_to_seconds(end_raw),

        "text":
            text.strip(),
    })


# ============================================================
# EXTRACT BORG ANNOTATIONS
# ============================================================

annotations = {}

log_lines = log_path.read_text(
    encoding="utf-8",
    errors="replace",
).splitlines()


for line in log_lines:

    fields = line.split("\t")

    if len(fields) < 5:
        continue

    timestamp = fields[0].strip()
    turn = fields[1].strip()
    speaker = fields[2].strip()
    event = fields[3].strip()
    value = "\t".join(fields[4:]).strip()

    if speaker != "BORG":
        continue

    if event not in {
        "BORG_CORRECTNESS",
        "BORG_CHECKED_UNCERTAIN",
    }:
        continue

    key = turn

    if key not in annotations:

        annotations[key] = {
            "turn": turn,
            "timestamp_sec":
                log_time_to_seconds(timestamp),

            "correct": None,
            "uncertain": None,
        }

    if event == "BORG_CORRECTNESS":

        annotations[key]["correct"] = value

    elif event == "BORG_CHECKED_UNCERTAIN":

        annotations[key]["uncertain"] = value


# ============================================================
# SHOW FIRST 15 ANNOTATIONS
# WITH NEAREST PRECEDING TRANSCRIPT
# ============================================================

print("=" * 100)
print("ALIGNMENT CHECK")
print("=" * 100)


for annotation in list(
    annotations.values()
)[:15]:

    annotation_time = (
        annotation["timestamp_sec"]
    )

    # Student utterances ending before the
    # BORG annotation timestamp
    candidates = [
        turn
        for turn in student_turns
        if turn["end_sec"] <= annotation_time
    ]

    if candidates:

        nearest = min(
            candidates,
            key=lambda x:
                annotation_time - x["end_sec"]
        )

        gap = (
            annotation_time
            - nearest["end_sec"]
        )

    else:

        nearest = None
        gap = None


    print("\n" + "-" * 100)

    print(
        "BORG turn:",
        annotation["turn"]
    )

    print(
        "BORG timestamp:",
        round(annotation_time, 3)
    )

    print(
        "Correct:",
        annotation["correct"]
    )

    print(
        "Uncertain:",
        annotation["uncertain"]
    )

    if nearest:

        print(
            "Nearest transcript:",
            nearest["text"]
        )

        print(
            "Transcript start:",
            round(nearest["start_sec"], 3)
        )

        print(
            "Transcript end:",
            round(nearest["end_sec"], 3)
        )

        print(
            "Gap after speech:",
            round(gap, 3),
            "seconds"
        )

    else:

        print(
            "No preceding transcript found."
        )
```

```python
from pathlib import Path
import pandas as pd


root = Path(
    "data/itspoke_uncertainty/PSLC-WOZ-allfiles"
)


# ============================================================
# TRANSCRIPT PARSER
# ============================================================

def read_transcript(path):

    rows = []

    for line in path.read_text(
        encoding="utf-8",
        errors="replace",
    ).splitlines():

        parts = line.split(",", 2)

        if len(parts) != 3:
            continue

        start, end, text = parts

        rows.append({
            "start": start.strip(),
            "end": end.strip(),
            "text": text.strip(),
        })

    return rows


# ============================================================
# BORG ANNOTATION PARSER
# ============================================================

def get_borg_annotations(log_path):

    annotations = {}

    for line in log_path.read_text(
        encoding="utf-8",
        errors="replace",
    ).splitlines():

        fields = line.split("\t")

        if len(fields) < 5:
            continue

        timestamp = fields[0].strip()
        turn = fields[1].strip()
        speaker = fields[2].strip()
        event = fields[3].strip()
        value = "\t".join(fields[4:]).strip()

        if speaker != "BORG":
            continue

        if event not in {
            "BORG_CORRECTNESS",
            "BORG_CHECKED_UNCERTAIN",
        }:
            continue

        if turn not in annotations:

            annotations[turn] = {
                "turn": turn,
                "timestamp_sec":
                    int(timestamp) / 1000,

                "correct": None,
                "uncertain": None,
            }

        if event == "BORG_CORRECTNESS":
            annotations[turn]["correct"] = value

        elif event == "BORG_CHECKED_UNCERTAIN":
            annotations[turn]["uncertain"] = value

    return list(annotations.values())


# ============================================================
# AUDIT ALL SUBJECTS
# ============================================================

audit_rows = []


transcript_dir = (
    root / "studenttranscripts-sos"
)

borg_dir = (
    root / "borglogs"
)


for transcript_path in sorted(
    transcript_dir.glob("pslc*.txt")
):

    student_id = transcript_path.stem

    log58 = (
        borg_dir
        / f"borg_{student_id}_58.log"
    )

    log59 = (
        borg_dir
        / f"borg_{student_id}_59.log"
    )

    if not log58.exists() or not log59.exists():
        print(
            "Missing log:",
            student_id
        )
        continue

    transcript_rows = read_transcript(
        transcript_path
    )

    ann58 = get_borg_annotations(
        log58
    )

    ann59 = get_borg_annotations(
        log59
    )

    total_annotations = (
        len(ann58)
        + len(ann59)
    )

    audit_rows.append({
        "student_id":
            student_id,

        "transcript_turns":
            len(transcript_rows),

        "problem58_annotations":
            len(ann58),

        "problem59_annotations":
            len(ann59),

        "total_annotations":
            total_annotations,

        "difference":
            len(transcript_rows)
            - total_annotations,
    })


audit_df = pd.DataFrame(
    audit_rows
)


print("=" * 90)
print("ITSPOKE TURN COUNT AUDIT")
print("=" * 90)

print(
    audit_df.to_string(
        index=False
    )
)


print("\n" + "=" * 90)
print("SUMMARY")
print("=" * 90)

print(
    "Subjects:",
    len(audit_df)
)

print(
    "Exact matches:",
    (
        audit_df["difference"] == 0
    ).sum()
)

print(
    "Mismatches:",
    (
        audit_df["difference"] != 0
    ).sum()
)


print("\nDifference distribution:")

print(
    audit_df[
        "difference"
    ].value_counts()
    .sort_index()
)


print("\nMismatching subjects:")

print(
    audit_df[
        audit_df["difference"] != 0
    ].to_string(
        index=False
    )
)
```

```python
from pathlib import Path
from datetime import datetime
import pandas as pd
import numpy as np


root = Path(
    "data/itspoke_uncertainty/PSLC-WOZ-allfiles"
)

transcript_dir = root / "studenttranscripts-sos"
borg_dir = root / "borglogs"


# ============================================================
# TIME FUNCTIONS
# ============================================================

def transcript_time_to_seconds(value):
    """
    0:00:48.682 -> 48.682 seconds
    """
    h, m, s = value.strip().split(":")

    return (
        int(h) * 3600
        + int(m) * 60
        + float(s)
    )


def parse_log_start_time(log_path):
    """
    Extract absolute timestamp such as:

    2007-01-10-01-28-13 -0500
    """

    for line in log_path.read_text(
        encoding="utf-8",
        errors="replace",
    ).splitlines():

        fields = line.split("\t")

        if (
            len(fields) >= 5
            and fields[3].strip() == "TIMESTAMP"
        ):
            value = fields[4].strip()

            return datetime.strptime(
                value,
                "%Y-%m-%d-%H-%M-%S %z"
            )

    raise ValueError(
        f"No TIMESTAMP header found in {log_path}"
    )


# ============================================================
# TRANSCRIPT
# ============================================================

def read_transcript(path):

    rows = []

    for line in path.read_text(
        encoding="utf-8",
        errors="replace",
    ).splitlines():

        parts = line.split(",", 2)

        if len(parts) != 3:
            continue

        start_raw, end_raw, text = parts

        rows.append({
            "start_sec":
                transcript_time_to_seconds(start_raw),

            "end_sec":
                transcript_time_to_seconds(end_raw),

            "student_text":
                text.strip(),
        })

    return rows


# ============================================================
# ANNOTATIONS
# ============================================================

def read_borg_annotations(
    log_path,
    problem_id,
    global_offset_sec,
):

    annotations = {}

    for line in log_path.read_text(
        encoding="utf-8",
        errors="replace",
    ).splitlines():

        fields = line.split("\t")

        if len(fields) < 5:
            continue

        timestamp_raw = fields[0].strip()
        turn = fields[1].strip()
        speaker = fields[2].strip()
        event = fields[3].strip()
        value = "\t".join(
            fields[4:]
        ).strip()

        if speaker != "BORG":
            continue

        if event not in {
            "BORG_CORRECTNESS",
            "BORG_CHECKED_UNCERTAIN",
        }:
            continue

        if turn not in annotations:

            local_sec = (
                int(timestamp_raw) / 1000
            )

            annotations[turn] = {
                "problem_id": problem_id,
                "turn_id": turn,

                "local_annotation_sec":
                    local_sec,

                "global_annotation_sec":
                    local_sec
                    + global_offset_sec,

                "correct": None,
                "uncertain": None,
            }

        if event == "BORG_CORRECTNESS":

            annotations[turn][
                "correct"
            ] = value.lower()

        elif event == "BORG_CHECKED_UNCERTAIN":

            annotations[turn][
                "uncertain"
            ] = value.lower()

    return list(
        annotations.values()
    )


# ============================================================
# MONOTONIC TIMESTAMP ALIGNMENT
#
# Annotation was observed approximately 4–6 seconds
# after speech in the verified example.
#
# Allow a generous 0–15 second matching window.
# ============================================================

def align_subject(
    transcript_rows,
    annotation_rows,
    target_gap=5.0,
    max_gap=15.0,
):

    matches = []

    used_transcripts = set()

    # Process annotations chronologically
    annotation_rows = sorted(
        annotation_rows,
        key=lambda x:
            x["global_annotation_sec"]
    )

    last_transcript_index = -1

    for annotation in annotation_rows:

        annotation_time = (
            annotation[
                "global_annotation_sec"
            ]
        )

        candidates = []

        for i, transcript in enumerate(
            transcript_rows
        ):

            if i <= last_transcript_index:
                continue

            gap = (
                annotation_time
                - transcript["end_sec"]
            )

            # Student speech should normally
            # finish shortly before annotation.
            if 0 <= gap <= max_gap:

                candidates.append(
                    (
                        abs(gap - target_gap),
                        gap,
                        i,
                        transcript,
                    )
                )

        if not candidates:

            matches.append({
                **annotation,

                "matched": False,

                "transcript_index": None,

                "start_sec": None,
                "end_sec": None,

                "student_text": None,

                "alignment_gap_sec": None,
            })

            continue

        # Prefer gap closest to observed ~5 sec
        candidates.sort(
            key=lambda x: x[0]
        )

        _, gap, index, transcript = (
            candidates[0]
        )

        used_transcripts.add(index)

        last_transcript_index = index

        matches.append({
            **annotation,

            "matched": True,

            "transcript_index": index,

            "start_sec":
                transcript["start_sec"],

            "end_sec":
                transcript["end_sec"],

            "student_text":
                transcript["student_text"],

            "alignment_gap_sec":
                gap,
        })

    return matches, used_transcripts


# ============================================================
# RUN ALL 60 SUBJECTS
# ============================================================

all_matches = []
subject_audit = []


for transcript_path in sorted(
    transcript_dir.glob("pslc*.txt")
):

    student_id = transcript_path.stem

    # condition is c1 / c2 / c3
    condition = student_id.split("-")[-1]

    log58 = (
        borg_dir
        / f"borg_{student_id}_58.log"
    )

    log59 = (
        borg_dir
        / f"borg_{student_id}_59.log"
    )

    if not log58.exists() or not log59.exists():

        print(
            "Missing logs for:",
            student_id
        )

        continue


    # --------------------------------------------------------
    # ABSOLUTE LOG START TIMES
    # --------------------------------------------------------

    start58 = parse_log_start_time(
        log58
    )

    start59 = parse_log_start_time(
        log59
    )

    problem59_offset = (
        start59 - start58
    ).total_seconds()


    # --------------------------------------------------------
    # READ DATA
    # --------------------------------------------------------

    transcript_rows = read_transcript(
        transcript_path
    )

    ann58 = read_borg_annotations(
        log58,
        problem_id=58,
        global_offset_sec=0,
    )

    ann59 = read_borg_annotations(
        log59,
        problem_id=59,
        global_offset_sec=problem59_offset,
    )

    annotations = ann58 + ann59


    # --------------------------------------------------------
    # ALIGN
    # --------------------------------------------------------

    matches, used_transcripts = (
        align_subject(
            transcript_rows,
            annotations,
            target_gap=5.0,
            max_gap=15.0,
        )
    )


    for match in matches:

        match["student_id"] = (
            student_id
        )

        match["condition"] = (
            condition
        )

        match["problem59_offset_sec"] = (
            problem59_offset
        )

        all_matches.append(
            match
        )


    matched_count = sum(
        m["matched"]
        for m in matches
    )

    unmatched_annotations = (
        len(matches)
        - matched_count
    )

    unused_transcripts = (
        len(transcript_rows)
        - len(used_transcripts)
    )


    matched_gaps = [
        m["alignment_gap_sec"]
        for m in matches
        if m["matched"]
    ]


    subject_audit.append({
        "student_id":
            student_id,

        "transcript_turns":
            len(transcript_rows),

        "annotations":
            len(annotations),

        "matched_annotations":
            matched_count,

        "unmatched_annotations":
            unmatched_annotations,

        "unused_transcripts":
            unused_transcripts,

        "problem59_offset_sec":
            problem59_offset,

        "mean_gap_sec":
            (
                np.mean(matched_gaps)
                if matched_gaps
                else np.nan
            ),

        "max_gap_sec":
            (
                np.max(matched_gaps)
                if matched_gaps
                else np.nan
            ),
    })


# ============================================================
# DATAFRAMES
# ============================================================

matches_df = pd.DataFrame(
    all_matches
)

alignment_audit_df = pd.DataFrame(
    subject_audit
)


# ============================================================
# SUMMARY
# ============================================================

print("=" * 100)
print("ITSPOKE TIMESTAMP ALIGNMENT AUDIT")
print("=" * 100)

total_annotations = len(
    matches_df
)

total_matched = (
    matches_df["matched"]
    .sum()
)

print(
    "Subjects:",
    len(alignment_audit_df)
)

print(
    "Total annotations:",
    total_annotations
)

print(
    "Matched annotations:",
    total_matched
)

print(
    "Unmatched annotations:",
    total_annotations
    - total_matched
)

print(
    "Overall match rate:",
    f"{100 * total_matched / total_annotations:.2f}%"
)


print("\nAlignment gap statistics:")

print(
    matches_df.loc[
        matches_df["matched"],
        "alignment_gap_sec"
    ].describe()
)


print("\n" + "=" * 100)
print("SUBJECTS WITH UNMATCHED ANNOTATIONS")
print("=" * 100)

problem_subjects = (
    alignment_audit_df[
        alignment_audit_df[
            "unmatched_annotations"
        ] > 0
    ]
)

print(
    problem_subjects.to_string(
        index=False
    )
)


print("\n" + "=" * 100)
print("UNMATCHED ANNOTATIONS")
print("=" * 100)

print(
    matches_df[
        ~matches_df["matched"]
    ][
        [
            "student_id",
            "problem_id",
            "turn_id",
            "global_annotation_sec",
            "correct",
            "uncertain",
        ]
    ].to_string(
        index=False
    )
)


# ============================================================
# SAVE AUDIT
# ============================================================

matches_df.to_csv(
    "itspoke_timestamp_alignment.csv",
    index=False,
)

alignment_audit_df.to_csv(
    "itspoke_timestamp_alignment_audit.csv",
    index=False,
)


print("\nSaved:")
print(
    "itspoke_timestamp_alignment.csv"
)

print(
    "itspoke_timestamp_alignment_audit.csv"
)
```

```python
import pandas as pd
import numpy as np

matches_df = pd.read_csv(
    "itspoke_timestamp_alignment.csv"
)

# ============================================================
# MATCHED DATA ONLY
# ============================================================

clean = matches_df[
    matches_df["matched"] == True
].copy()

print("=" * 90)
print("ITSPOKE MATCHED-DATA AUDIT")
print("=" * 90)

print("Matched rows:", len(clean))


# ============================================================
# CONVERT LABELS TO BOOLEAN / 0-1
# ============================================================

def parse_bool(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value == "true":
        return 1

    if value == "false":
        return 0

    return np.nan


clean["uncertain_label"] = (
    clean["uncertain"]
    .apply(parse_bool)
)

clean["correct_label"] = (
    clean["correct"]
    .apply(parse_bool)
)


# ============================================================
# LABEL COMPLETENESS
# ============================================================

print("\n" + "=" * 90)
print("LABEL COMPLETENESS")
print("=" * 90)

print(
    clean[
        [
            "uncertain_label",
            "correct_label",
        ]
    ]
    .isna()
    .sum()
)


# ============================================================
# UNCERTAINTY DISTRIBUTION
# ============================================================

print("\n" + "=" * 90)
print("UNCERTAINTY DISTRIBUTION")
print("=" * 90)

print(
    clean[
        "uncertain_label"
    ]
    .value_counts(
        dropna=False
    )
)

print("\nPercentages:")

print(
    clean[
        "uncertain_label"
    ]
    .value_counts(
        normalize=True,
        dropna=False
    )
    .mul(100)
    .round(2)
)


# ============================================================
# CORRECTNESS DISTRIBUTION
# ============================================================

print("\n" + "=" * 90)
print("CORRECTNESS DISTRIBUTION")
print("=" * 90)

print(
    clean[
        "correct_label"
    ]
    .value_counts(
        dropna=False
    )
)


# ============================================================
# CORRECTNESS × UNCERTAINTY
# ============================================================

print("\n" + "=" * 90)
print("CORRECTNESS × UNCERTAINTY")
print("=" * 90)

cross = pd.crosstab(
    clean["correct_label"],
    clean["uncertain_label"],
    margins=True,
)

print(cross)


# ============================================================
# ALIGNMENT GAP BANDS
# ============================================================

print("\n" + "=" * 90)
print("ALIGNMENT GAP QUALITY")
print("=" * 90)

for threshold in [
    6,
    8,
    10,
    12,
    15,
]:

    count = (
        clean[
            "alignment_gap_sec"
        ] <= threshold
    ).sum()

    pct = (
        count / len(clean)
    ) * 100

    print(
        f"≤ {threshold:2d} sec : "
        f"{count:4d} "
        f"({pct:6.2f}%)"
    )


# ============================================================
# GAP BINS
# ============================================================

clean["gap_band"] = pd.cut(
    clean["alignment_gap_sec"],
    bins=[
        0,
        3,
        6,
        10,
        15,
    ],
    include_lowest=True,
)

print("\nGap bands:")

print(
    clean[
        "gap_band"
    ]
    .value_counts()
    .sort_index()
)


# ============================================================
# SUBJECT COVERAGE
# ============================================================

print("\n" + "=" * 90)
print("SUBJECT COVERAGE")
print("=" * 90)

print(
    "Unique students:",
    clean["student_id"].nunique()
)

print("\nTurns per student:")

print(
    clean.groupby(
        "student_id"
    )
    .size()
    .describe()
)


# ============================================================
# CONDITION DISTRIBUTION
# ============================================================

print("\nCondition:")

print(
    clean[
        "condition"
    ]
    .value_counts()
)


# ============================================================
# PROBLEM DISTRIBUTION
# ============================================================

print("\nProblem:")

print(
    clean[
        "problem_id"
    ]
    .value_counts()
)


# ============================================================
# EXAMPLE UNCERTAIN TURNS
# ============================================================

print("\n" + "=" * 90)
print("RANDOM UNCERTAIN EXAMPLES")
print("=" * 90)

uncertain_examples = (
    clean[
        clean[
            "uncertain_label"
        ] == 1
    ]
    .sample(
        n=min(
            20,
            (
                clean[
                    "uncertain_label"
                ] == 1
            ).sum()
        ),
        random_state=42,
    )
)

for _, row in uncertain_examples.iterrows():

    print(
        f"\nStudent: {row['student_id']}"
    )

    print(
        f"Problem: {row['problem_id']}"
    )

    print(
        f"Gap: "
        f"{row['alignment_gap_sec']:.3f}s"
    )

    print(
        f"Correct: "
        f"{row['correct_label']}"
    )

    print(
        "Text:",
        row["student_text"]
    )

    print("-" * 70)


# ============================================================
# SAVE CLEAN MATCHED DATA
# ============================================================

clean.to_csv(
    "itspoke_matched_clean.csv",
    index=False,
)

print(
    "\nSaved: "
    "itspoke_matched_clean.csv"
)
```

```python
import pandas as pd
import re


df = pd.read_csv(
    "itspoke_matched_clean.csv"
)

data = df[
    df["alignment_gap_sec"] <= 10
].copy()


def normalize_itspoke_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # --------------------------------------------
    # Remove transcription convention tags
    # <nspn>, <spn>, <X>, etc.
    # --------------------------------------------
    text = re.sub(
        r"<[^>]+>",
        " ",
        text,
    )

    # --------------------------------------------
    # Remove overlap markers
    # ***word*** -> word
    # --------------------------------------------
    text = text.replace(
        "***",
        ""
    )

    # --------------------------------------------
    # False-start notation:
    # grav+gravity -> grav gravity
    # --------------------------------------------
    text = text.replace(
        "+",
        " "
    )

    # --------------------------------------------
    # Remove punctuation supplied by transcriber.
    #
    # This deliberately removes:
    # ?, ??, ..., commas, etc.
    # --------------------------------------------
    text = re.sub(
        r"[^\w\s']",
        " ",
        text,
    )

    # --------------------------------------------
    # Normalize whitespace
    # --------------------------------------------
    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


data["text_raw"] = (
    data["student_text"]
    .fillna("")
)

data["text_normalized"] = (
    data["student_text"]
    .apply(normalize_itspoke_text)
)


print("=" * 80)
print("NORMALIZATION EXAMPLES")
print("=" * 80)

sample = data.sample(
    n=20,
    random_state=42,
)

for _, row in sample.iterrows():

    print("\nRAW:")
    print(row["text_raw"])

    print("NORMALIZED:")
    print(row["text_normalized"])

    print(
        "UNCERTAIN:",
        row["uncertain_label"]
    )

    print("-" * 70)


print("\nEmpty normalized texts:")

print(
    (
        data["text_normalized"]
        .str.len() == 0
    ).sum()
)
```

```python
from sklearn.model_selection import (
    StratifiedGroupKFold,
)

# ============================================================
# OUTER SPLIT:
# approximately 80% development / 20% final test
# ============================================================

outer = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42,
)

outer_splits = list(
    outer.split(
        data,
        y=data["uncertain_label"],
        groups=data["student_id"],
    )
)

dev_idx, test_idx = outer_splits[0]

development = (
    data.iloc[dev_idx]
    .copy()
)

final_test = (
    data.iloc[test_idx]
    .copy()
)


# ============================================================
# DEVELOPMENT SPLIT:
# train / validation
# ============================================================

inner = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=43,
)

inner_splits = list(
    inner.split(
        development,
        y=development["uncertain_label"],
        groups=development["student_id"],
    )
)

train_idx, val_idx = inner_splits[0]

train = (
    development.iloc[train_idx]
    .copy()
)

validation = (
    development.iloc[val_idx]
    .copy()
)


print("=" * 90)
print("ITSPOKE SUBJECT-DISJOINT SPLIT")
print("=" * 90)


def describe_split(
    name,
    split,
):

    print(f"\n{name}")

    print(
        "Rows:",
        len(split)
    )

    print(
        "Students:",
        split[
            "student_id"
        ].nunique()
    )

    print(
        "Uncertain:",
        split[
            "uncertain_label"
        ].sum()
    )

    print(
        "Uncertainty rate:",
        round(
            split[
                "uncertain_label"
            ].mean(),
            4,
        )
    )

    print(
        "Conditions:",
        split[
            "condition"
        ].value_counts()
        .to_dict()
    )


describe_split(
    "TRAIN",
    train,
)

describe_split(
    "VALIDATION",
    validation,
)

describe_split(
    "FINAL TEST",
    final_test,
)


# ============================================================
# LEAKAGE CHECK
# ============================================================

train_students = set(
    train["student_id"]
)

val_students = set(
    validation["student_id"]
)

test_students = set(
    final_test["student_id"]
)


print("\n" + "=" * 90)
print("STUDENT LEAKAGE CHECK")
print("=" * 90)

print(
    "Train ∩ validation:",
    len(
        train_students
        & val_students
    )
)

print(
    "Train ∩ test:",
    len(
        train_students
        & test_students
    )
)

print(
    "Validation ∩ test:",
    len(
        val_students
        & test_students
    )
)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split


# ============================================================
# STUDENT-LEVEL SUMMARY
# ============================================================

student_summary = (
    data
    .groupby(
        [
            "student_id",
            "condition",
        ],
        as_index=False,
    )
    .agg(
        n_turns=(
            "uncertain_label",
            "size",
        ),
        uncertain_turns=(
            "uncertain_label",
            "sum",
        ),
        uncertainty_rate=(
            "uncertain_label",
            "mean",
        ),
    )
)


print("=" * 90)
print("STUDENT SUMMARY")
print("=" * 90)

print(
    student_summary
    .groupby("condition")
    .size()
)

print("\nOverall student uncertainty rates:")

print(
    student_summary[
        "uncertainty_rate"
    ].describe()
)


# ============================================================
# BALANCED CONDITION-AWARE SPLIT
#
# For each condition:
#
# 20 students
# → 12 train
# → 4 validation
# → 4 final test
#
# We sort students by uncertainty rate and distribute
# them cyclically across four blocks to reduce imbalance.
# ============================================================

rng = np.random.RandomState(42)

train_students = []
val_students = []
test_students = []


for condition in [
    "c1",
    "c2",
    "c3",
]:

    subset = (
        student_summary[
            student_summary[
                "condition"
            ] == condition
        ]
        .copy()
    )

    # Slight random jitter avoids deterministic
    # ordering when students have identical rates.
    subset["jitter"] = rng.uniform(
        0,
        1e-6,
        size=len(subset),
    )

    subset = subset.sort_values(
        [
            "uncertainty_rate",
            "jitter",
        ]
    ).reset_index(drop=True)


    # Divide ordered students into 4 roughly
    # comparable strata by uncertainty rate.
    subset["rate_bin"] = pd.qcut(
        subset["uncertainty_rate"],
        q=4,
        labels=False,
        duplicates="drop",
    )


    condition_train = []
    condition_val = []
    condition_test = []


    for _, group in subset.groupby(
        "rate_bin"
    ):

        ids = (
            group[
                "student_id"
            ]
            .sample(
                frac=1,
                random_state=42,
            )
            .tolist()
        )

        # Roughly 60 / 20 / 20 within
        # every uncertainty-rate bin.
        n = len(ids)

        n_test = max(
            1,
            round(n * 0.20),
        )

        n_val = max(
            1,
            round(n * 0.20),
        )

        condition_test.extend(
            ids[:n_test]
        )

        condition_val.extend(
            ids[
                n_test:
                n_test + n_val
            ]
        )

        condition_train.extend(
            ids[
                n_test + n_val:
            ]
        )


    # --------------------------------------------------------
    # Force exact 12 / 4 / 4 per condition
    # --------------------------------------------------------

    all_ids = subset[
        "student_id"
    ].tolist()

    current_test = list(
        dict.fromkeys(
            condition_test
        )
    )

    current_val = list(
        dict.fromkeys(
            condition_val
        )
    )

    # Fill / trim test
    current_test = current_test[:4]

    remaining = [
        x
        for x in all_ids
        if x not in current_test
    ]

    while len(current_test) < 4:
        current_test.append(
            remaining.pop(0)
        )


    # Fill / trim validation
    current_val = [
        x
        for x in current_val
        if x not in current_test
    ][:4]

    remaining = [
        x
        for x in all_ids
        if (
            x not in current_test
            and x not in current_val
        )
    ]

    while len(current_val) < 4:
        current_val.append(
            remaining.pop(0)
        )


    current_train = [
        x
        for x in all_ids
        if (
            x not in current_test
            and x not in current_val
        )
    ]


    assert len(
        current_train
    ) == 12

    assert len(
        current_val
    ) == 4

    assert len(
        current_test
    ) == 4


    train_students.extend(
        current_train
    )

    val_students.extend(
        current_val
    )

    test_students.extend(
        current_test
    )


# ============================================================
# CREATE TURN-LEVEL DATASETS
# ============================================================

train = data[
    data["student_id"]
    .isin(train_students)
].copy()

validation = data[
    data["student_id"]
    .isin(val_students)
].copy()

final_test = data[
    data["student_id"]
    .isin(test_students)
].copy()


# ============================================================
# AUDIT
# ============================================================

def describe_split(
    name,
    split,
):

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(
        "Rows:",
        len(split)
    )

    print(
        "Students:",
        split[
            "student_id"
        ].nunique()
    )

    print(
        "Uncertain:",
        split[
            "uncertain_label"
        ].sum()
    )

    print(
        "Uncertainty rate:",
        round(
            split[
                "uncertain_label"
            ].mean(),
            4,
        )
    )

    print(
        "\nStudents by condition:"
    )

    print(
        split[
            [
                "student_id",
                "condition",
            ]
        ]
        .drop_duplicates()
        ["condition"]
        .value_counts()
    )

    print(
        "\nTurns by condition:"
    )

    print(
        split[
            "condition"
        ]
        .value_counts()
    )


describe_split(
    "TRAIN",
    train,
)

describe_split(
    "VALIDATION",
    validation,
)

describe_split(
    "FINAL TEST",
    final_test,
)


# ============================================================
# LEAKAGE CHECK
# ============================================================

train_ids = set(
    train["student_id"]
)

val_ids = set(
    validation["student_id"]
)

test_ids = set(
    final_test["student_id"]
)


print("\n" + "=" * 90)
print("LEAKAGE CHECK")
print("=" * 90)

print(
    "Train ∩ Validation:",
    len(train_ids & val_ids)
)

print(
    "Train ∩ Test:",
    len(train_ids & test_ids)
)

print(
    "Validation ∩ Test:",
    len(val_ids & test_ids)
)


# ============================================================
# SAVE ONLY AFTER INSPECTING OUTPUT
# ============================================================

# train.to_csv(
#     "itspoke_train.csv",
#     index=False,
# )
#
# validation.to_csv(
#     "itspoke_validation.csv",
#     index=False,
# )
#
# final_test.to_csv(
#     "itspoke_final_test.csv",
#     index=False,
# )
```

```python
import numpy as np
import pandas as pd


# ============================================================
# TARGETS
# ============================================================

GLOBAL_RATE = data["uncertain_label"].mean()
TOTAL_ROWS = len(data)

TARGET_TRAIN_ROWS = TOTAL_ROWS * 0.60
TARGET_VAL_ROWS = TOTAL_ROWS * 0.20
TARGET_TEST_ROWS = TOTAL_ROWS * 0.20


print("Global uncertainty rate:", GLOBAL_RATE)
print("Total rows:", TOTAL_ROWS)


# ============================================================
# STUDENT SUMMARY
# ============================================================

student_summary = (
    data
    .groupby(
        ["student_id", "condition"],
        as_index=False
    )
    .agg(
        n_turns=("uncertain_label", "size"),
        uncertain_turns=("uncertain_label", "sum"),
    )
)

student_summary["uncertainty_rate"] = (
    student_summary["uncertain_turns"]
    / student_summary["n_turns"]
)


# ============================================================
# SCORE ONE SPLIT
# ============================================================

def score_split(
    train_ids,
    val_ids,
    test_ids,
):

    train_part = student_summary[
        student_summary["student_id"].isin(train_ids)
    ]

    val_part = student_summary[
        student_summary["student_id"].isin(val_ids)
    ]

    test_part = student_summary[
        student_summary["student_id"].isin(test_ids)
    ]


    def stats(part):

        rows = part["n_turns"].sum()
        uncertain = part["uncertain_turns"].sum()

        rate = (
            uncertain / rows
            if rows > 0
            else 0
        )

        return rows, rate


    train_rows, train_rate = stats(train_part)
    val_rows, val_rate = stats(val_part)
    test_rows, test_rate = stats(test_part)


    # --------------------------------------------------------
    # LABEL-BALANCE PENALTY
    #
    # This is the most important part.
    # --------------------------------------------------------

    rate_error = (
        abs(train_rate - GLOBAL_RATE)
        +
        abs(val_rate - GLOBAL_RATE)
        +
        abs(test_rate - GLOBAL_RATE)
    )


    # --------------------------------------------------------
    # DATASET-SIZE PENALTY
    # --------------------------------------------------------

    row_error = (
        abs(train_rows - TARGET_TRAIN_ROWS)
        / TOTAL_ROWS
        +
        abs(val_rows - TARGET_VAL_ROWS)
        / TOTAL_ROWS
        +
        abs(test_rows - TARGET_TEST_ROWS)
        / TOTAL_ROWS
    )


    # Give label balance greater importance
    score = (
        5.0 * rate_error
        +
        1.0 * row_error
    )


    return {
        "score": score,

        "train_rows": train_rows,
        "val_rows": val_rows,
        "test_rows": test_rows,

        "train_rate": train_rate,
        "val_rate": val_rate,
        "test_rate": test_rate,
    }


# ============================================================
# RANDOM SEARCH
# ============================================================

rng = np.random.default_rng(42)

BEST = None

N_SEARCH = 100_000


for iteration in range(N_SEARCH):

    train_ids = []
    val_ids = []
    test_ids = []


    for condition in ["c1", "c2", "c3"]:

        ids = (
            student_summary[
                student_summary["condition"] == condition
            ]["student_id"]
            .to_numpy()
            .copy()
        )

        rng.shuffle(ids)

        # Exactly 12 / 4 / 4 per condition
        train_ids.extend(
            ids[:12]
        )

        val_ids.extend(
            ids[12:16]
        )

        test_ids.extend(
            ids[16:20]
        )


    stats = score_split(
        train_ids,
        val_ids,
        test_ids,
    )


    if (
        BEST is None
        or stats["score"] < BEST["score"]
    ):

        BEST = {
            **stats,
            "train_ids": list(train_ids),
            "val_ids": list(val_ids),
            "test_ids": list(test_ids),
            "iteration": iteration,
        }


# ============================================================
# BEST RESULT
# ============================================================

print("\n" + "=" * 90)
print("BEST STUDENT-DISJOINT SPLIT")
print("=" * 90)

print(
    "Found at iteration:",
    BEST["iteration"]
)

print(
    "Score:",
    round(BEST["score"], 6)
)

print("\nRows:")

print(
    "Train      :",
    BEST["train_rows"]
)

print(
    "Validation :",
    BEST["val_rows"]
)

print(
    "Test       :",
    BEST["test_rows"]
)


print("\nUncertainty rates:")

print(
    "Full       :",
    round(GLOBAL_RATE, 4)
)

print(
    "Train      :",
    round(BEST["train_rate"], 4)
)

print(
    "Validation :",
    round(BEST["val_rate"], 4)
)

print(
    "Test       :",
    round(BEST["test_rate"], 4)
)


# ============================================================
# CREATE TURN-LEVEL SPLITS
# ============================================================

train = data[
    data["student_id"].isin(
        BEST["train_ids"]
    )
].copy()

validation = data[
    data["student_id"].isin(
        BEST["val_ids"]
    )
].copy()

final_test = data[
    data["student_id"].isin(
        BEST["test_ids"]
    )
].copy()


# ============================================================
# FINAL AUDIT
# ============================================================

def audit_split(name, split):

    students = (
        split[
            ["student_id", "condition"]
        ]
        .drop_duplicates()
    )

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(
        "Rows:",
        len(split)
    )

    print(
        "Students:",
        split["student_id"].nunique()
    )

    print(
        "Uncertain:",
        split["uncertain_label"].sum()
    )

    print(
        "Uncertainty rate:",
        round(
            split["uncertain_label"].mean(),
            4,
        )
    )

    print("\nStudents by condition:")

    print(
        students["condition"]
        .value_counts()
        .sort_index()
    )


audit_split(
    "TRAIN",
    train
)

audit_split(
    "VALIDATION",
    validation
)

audit_split(
    "FINAL TEST",
    final_test
)


# ============================================================
# LEAKAGE
# ============================================================

train_set = set(train["student_id"])
val_set = set(validation["student_id"])
test_set = set(final_test["student_id"])

print("\n" + "=" * 90)
print("LEAKAGE CHECK")
print("=" * 90)

print(
    "Train ∩ Validation:",
    len(train_set & val_set)
)

print(
    "Train ∩ Test:",
    len(train_set & test_set)
)

print(
    "Validation ∩ Test:",
    len(val_set & test_set)
)
```

```python
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
)


# ============================================================
# MODEL FACTORY
# ============================================================

def build_model(class_weight=None):

    return Pipeline([
        (
            "tfidf",
            TfidfVectorizer(
                lowercase=True,
                ngram_range=(1, 2),
                min_df=2,
                max_df=0.95,
                sublinear_tf=True,
            ),
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                random_state=42,
                class_weight=class_weight,
            ),
        ),
    ])


# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model_name,
    model,
    text_column,
):

    model.fit(
        train[text_column],
        train["uncertain_label"],
    )

    prob = model.predict_proba(
        validation[text_column]
    )[:, 1]

    pred_05 = (
        prob >= 0.50
    ).astype(int)

    print("\n" + "=" * 90)
    print(model_name)
    print("=" * 90)

    print(
        "Accuracy:",
        round(
            accuracy_score(
                validation["uncertain_label"],
                pred_05,
            ),
            4,
        )
    )

    print(
        "Balanced accuracy:",
        round(
            balanced_accuracy_score(
                validation["uncertain_label"],
                pred_05,
            ),
            4,
        )
    )

    print(
        "Precision:",
        round(
            precision_score(
                validation["uncertain_label"],
                pred_05,
                zero_division=0,
            ),
            4,
        )
    )

    print(
        "Recall:",
        round(
            recall_score(
                validation["uncertain_label"],
                pred_05,
                zero_division=0,
            ),
            4,
        )
    )

    print(
        "F1:",
        round(
            f1_score(
                validation["uncertain_label"],
                pred_05,
                zero_division=0,
            ),
            4,
        )
    )

    print(
        "PR-AUC:",
        round(
            average_precision_score(
                validation["uncertain_label"],
                prob,
            ),
            4,
        )
    )

    print("\nConfusion matrix:")
    print(
        confusion_matrix(
            validation["uncertain_label"],
            pred_05,
        )
    )


    # ========================================================
    # THRESHOLD SWEEP
    # ========================================================

    rows = []

    for threshold in np.arange(
        0.05,
        0.96,
        0.05,
    ):

        pred = (
            prob >= threshold
        ).astype(int)

        rows.append({
            "threshold":
                round(threshold, 2),

            "precision":
                precision_score(
                    validation["uncertain_label"],
                    pred,
                    zero_division=0,
                ),

            "recall":
                recall_score(
                    validation["uncertain_label"],
                    pred,
                    zero_division=0,
                ),

            "f1":
                f1_score(
                    validation["uncertain_label"],
                    pred,
                    zero_division=0,
                ),

            "balanced_accuracy":
                balanced_accuracy_score(
                    validation["uncertain_label"],
                    pred,
                ),

            "predicted_positive":
                pred.sum(),
        })


    thresholds = pd.DataFrame(rows)

    best_f1 = thresholds.loc[
        thresholds["f1"].idxmax()
    ]

    best_balanced = thresholds.loc[
        thresholds[
            "balanced_accuracy"
        ].idxmax()
    ]

    print("\nBest F1 threshold:")
    print(best_f1)

    print("\nBest balanced-accuracy threshold:")
    print(best_balanced)

    return {
        "name": model_name,
        "model": model,
        "prob": prob,
        "thresholds": thresholds,
        "pr_auc":
            average_precision_score(
                validation[
                    "uncertain_label"
                ],
                prob,
            ),
        "best_f1_threshold":
            best_f1["threshold"],
        "best_f1":
            best_f1["f1"],
        "best_balanced_threshold":
            best_balanced["threshold"],
        "best_balanced_accuracy":
            best_balanced[
                "balanced_accuracy"
            ],
    }


# ============================================================
# A. NORMALIZED, UNWEIGHTED
# ============================================================

result_A = evaluate_model(
    "A. NORMALIZED / UNWEIGHTED",
    build_model(
        class_weight=None
    ),
    "text_normalized",
)


# ============================================================
# B. NORMALIZED, CLASS-WEIGHTED
# ============================================================

result_B = evaluate_model(
    "B. NORMALIZED / BALANCED",
    build_model(
        class_weight="balanced"
    ),
    "text_normalized",
)


# ============================================================
# C. RAW TRANSCRIPT ABLATION
# ============================================================

result_C = evaluate_model(
    "C. RAW TRANSCRIPT / UNWEIGHTED",
    build_model(
        class_weight=None
    ),
    "text_raw",
)


# ============================================================
# SUMMARY
# ============================================================

summary = pd.DataFrame([
    {
        "model":
            r["name"],

        "PR_AUC":
            r["pr_auc"],

        "best_F1":
            r["best_f1"],

        "best_F1_threshold":
            r["best_f1_threshold"],

        "best_balanced_accuracy":
            r[
                "best_balanced_accuracy"
            ],

        "best_balanced_threshold":
            r[
                "best_balanced_threshold"
            ],
    }

    for r in [
        result_A,
        result_B,
        result_C,
    ]
])


print("\n" + "=" * 90)
print("MODEL COMPARISON")
print("=" * 90)

print(
    summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}",
    )
)
```
