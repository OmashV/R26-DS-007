```python
import json
from pathlib import Path

path = Path("data/cima/dataset.json")

with path.open(
    "r",
    encoding="utf-8"
) as f:
    cima = json.load(f)


print("=" * 80)
print("CIMA DATASET AUDIT")
print("=" * 80)

print("Top-level type:")
print(type(cima))

print("\nTop-level keys:")
print(list(cima.keys()))


for dataset_name, dataset_part in cima.items():

    print("\n" + "=" * 80)
    print("SECTION:", dataset_name)
    print("=" * 80)

    print("Type:", type(dataset_part))
    print("Length:", len(dataset_part))

    # --------------------------------------------------------
    # EMPTY SECTION
    # --------------------------------------------------------

    if len(dataset_part) == 0:

        print("This section is empty — skipping.")
        continue


    # --------------------------------------------------------
    # DICTIONARY SECTION
    # --------------------------------------------------------

    if isinstance(dataset_part, dict):

        first_key = next(
            iter(dataset_part)
        )

        first = dataset_part[
            first_key
        ]

        print(
            "\nFirst key:",
            first_key
        )

        print(
            "\nFirst item type:",
            type(first)
        )

        if isinstance(first, dict):

            print(
                "\nFields:"
            )

            print(
                list(first.keys())
            )

            print(
                "\nFirst example:"
            )

            print(first)

        else:

            print(
                "\nFirst item:"
            )

            print(first)


    # --------------------------------------------------------
    # LIST SECTION
    # --------------------------------------------------------

    elif isinstance(dataset_part, list):

        first = dataset_part[0]

        print(
            "\nFirst item type:",
            type(first)
        )

        if isinstance(first, dict):

            print(
                "\nFields:"
            )

            print(
                list(first.keys())
            )

        print(
            "\nFirst item:"
        )

        print(first)


    else:

        print(
            "\nValue:"
        )

        print(dataset_part)
```

```python
from collections import Counter

prep = cima["prepDataset"]

STUDENT_ACTIONS = [
    "guess",
    "clarification_question",
    "affirmation",
    "other",
]


def to_bool(value):
    return (
        value is True
        or str(value).strip().lower() == "true"
    )


counts = Counter()

multi_label_count = 0
zero_label_count = 0
invalid_count = 0

total = 0


for example_id, example in prep.items():

    actions = example.get(
        "studentActions"
    )

    if (
        actions is None
        or len(actions) != 4
    ):
        invalid_count += 1
        continue

    bool_actions = [
        to_bool(x)
        for x in actions
    ]

    total += 1

    positives = sum(bool_actions)

    if positives > 1:
        multi_label_count += 1

    if positives == 0:
        zero_label_count += 1

    for name, value in zip(
        STUDENT_ACTIONS,
        bool_actions
    ):

        if value:
            counts[name] += 1


print("=" * 80)
print("CIMA STUDENT ACTION DISTRIBUTION")
print("=" * 80)

print("Valid examples:", total)

for name in STUDENT_ACTIONS:

    count = counts[name]

    pct = (
        count / total * 100
        if total
        else 0
    )

    print(
        f"{name:25s}"
        f"{count:5d}"
        f" ({pct:6.2f}%)"
    )


print(
    "\nMulti-label examples:",
    multi_label_count
)

print(
    "Zero-label examples:",
    zero_label_count
)

print(
    "Invalid examples:",
    invalid_count
)
```

```python
import pandas as pd


rows = []


for example_id, example in prep.items():

    past_convo = example.get(
        "past_convo",
        []
    )

    actions = example.get(
        "studentActions"
    )

    if (
        not past_convo
        or actions is None
        or len(actions) != 4
    ):
        continue

    bool_actions = [
        to_bool(x)
        for x in actions
    ]


    # Last past_convo entry is the
    # target student utterance.
    target = str(
        past_convo[-1]
    ).strip()


    # Everything before it is dialogue context.
    context_turns = [
        str(x).strip()
        for x in past_convo[:-1]
    ]


    rows.append({
        "example_id":
            str(example_id),

        "target_text":
            target,

        "context_text":
            "\n".join(
                context_turns
            ),

        "guess":
            int(bool_actions[0]),

        "clarification_request":
            int(bool_actions[1]),

        "affirmation":
            int(bool_actions[2]),

        "other":
            int(bool_actions[3]),

        "num_context_turns":
            len(context_turns),
    })


cima_df = pd.DataFrame(rows)


print("=" * 80)
print("CIMA CLEAN DATAFRAME")
print("=" * 80)

print("Shape:", cima_df.shape)

print("\nColumns:")
print(cima_df.columns.tolist())

print("\nClarification distribution:")
print(
    cima_df[
        "clarification_request"
    ].value_counts()
)

print("\nPercentage:")
print(
    cima_df[
        "clarification_request"
    ]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nMissing target text:")
print(
    cima_df[
        "target_text"
    ].isna().sum()
)

print("\nEmpty target text:")
print(
    (
        cima_df[
            "target_text"
        ].str.strip() == ""
    ).sum()
)
```

```python
import pandas as pd
import re


# ============================================================
# NORMALIZE TARGET FOR DUPLICATE AUDIT
# ============================================================

def normalize_target(text):

    text = str(text).lower().strip()

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Remove punctuation for duplicate comparison
    text = re.sub(
        r"[^\w\s']",
        "",
        text
    )

    return text.strip()


cima_df["target_normalized"] = (
    cima_df["target_text"]
    .apply(normalize_target)
)


print("=" * 90)
print("CIMA DUPLICATE / LABEL AUDIT")
print("=" * 90)

print(
    "Total examples:",
    len(cima_df)
)

print(
    "Unique raw targets:",
    cima_df["target_text"].nunique()
)

print(
    "Unique normalized targets:",
    cima_df["target_normalized"].nunique()
)

print(
    "Duplicate normalized target rows:",
    cima_df[
        "target_normalized"
    ].duplicated().sum()
)


# ============================================================
# LABEL CONFLICTS
#
# Same normalized utterance appears with both
# clarification=0 and clarification=1.
# ============================================================

label_sets = (
    cima_df
    .groupby(
        "target_normalized"
    )["clarification_request"]
    .agg(
        lambda x:
            tuple(sorted(set(x)))
    )
)


conflicting_targets = label_sets[
    label_sets.apply(
        lambda x: len(x) > 1
    )
]


print(
    "\nNormalized targets with conflicting labels:",
    len(conflicting_targets)
)


# ============================================================
# MOST FREQUENT TARGETS
# ============================================================

print("\n" + "=" * 90)
print("MOST FREQUENT NORMALIZED TARGETS")
print("=" * 90)

print(
    cima_df[
        "target_normalized"
    ]
    .value_counts()
    .head(30)
)


# ============================================================
# SHOW CONFLICT EXAMPLES
# ============================================================

print("\n" + "=" * 90)
print("SAMPLE TARGETS WITH LABEL CONFLICTS")
print("=" * 90)

for target in list(
    conflicting_targets.index
)[:20]:

    subset = cima_df[
        cima_df[
            "target_normalized"
        ] == target
    ]

    print("\nTARGET:", target)

    print(
        subset[
            [
                "target_text",
                "clarification_request",
                "guess",
                "affirmation",
            ]
        ].to_string(
            index=False
        )
    )

    print("-" * 80)


# ============================================================
# MULTI-LABEL OVERLAP
# ============================================================

print("\n" + "=" * 90)
print("CLARIFICATION LABEL OVERLAP")
print("=" * 90)

clarification_rows = cima_df[
    cima_df[
        "clarification_request"
    ] == 1
]

print(
    "Clarification total:",
    len(clarification_rows)
)

print(
    "Clarification + Guess:",
    (
        (
            clarification_rows["guess"] == 1
        )
    ).sum()
)

print(
    "Clarification + Affirmation:",
    (
        (
            clarification_rows["affirmation"] == 1
        )
    ).sum()
)

print(
    "Clarification + Other:",
    (
        (
            clarification_rows["other"] == 1
        )
    ).sum()
)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit


# ============================================================
# BASIC TARGET
# ============================================================

GLOBAL_RATE = (
    cima_df["clarification_request"].mean()
)

TOTAL_ROWS = len(cima_df)

TARGET_TRAIN = TOTAL_ROWS * 0.60
TARGET_VAL = TOTAL_ROWS * 0.20
TARGET_TEST = TOTAL_ROWS * 0.20


print(
    "Global clarification rate:",
    round(GLOBAL_RATE, 4)
)

print(
    "Total rows:",
    TOTAL_ROWS
)

print(
    "Unique target groups:",
    cima_df["target_normalized"].nunique()
)


# ============================================================
# TRY MANY GROUP-DISJOINT SPLITS
# ============================================================

best = None


for seed in range(5000):

    # --------------------------------------------------------
    # OUTER:
    # 80% development / 20% final test
    # --------------------------------------------------------

    outer = GroupShuffleSplit(
        n_splits=1,
        test_size=0.20,
        random_state=seed,
    )

    dev_idx, test_idx = next(
        outer.split(
            cima_df,
            y=cima_df[
                "clarification_request"
            ],
            groups=cima_df[
                "target_normalized"
            ],
        )
    )

    development = (
        cima_df.iloc[dev_idx]
        .copy()
    )

    test = (
        cima_df.iloc[test_idx]
        .copy()
    )


    # --------------------------------------------------------
    # INNER:
    # 75% of development -> train
    # 25% of development -> validation
    #
    # Gives approximately:
    # 60% train / 20% val / 20% test
    # --------------------------------------------------------

    inner = GroupShuffleSplit(
        n_splits=1,
        test_size=0.25,
        random_state=seed + 10000,
    )

    train_idx, val_idx = next(
        inner.split(
            development,
            y=development[
                "clarification_request"
            ],
            groups=development[
                "target_normalized"
            ],
        )
    )

    train_part = (
        development.iloc[train_idx]
        .copy()
    )

    val_part = (
        development.iloc[val_idx]
        .copy()
    )


    # --------------------------------------------------------
    # SCORE SPLIT
    # --------------------------------------------------------

    train_rate = (
        train_part[
            "clarification_request"
        ].mean()
    )

    val_rate = (
        val_part[
            "clarification_request"
        ].mean()
    )

    test_rate = (
        test[
            "clarification_request"
        ].mean()
    )


    rate_error = (
        abs(train_rate - GLOBAL_RATE)
        +
        abs(val_rate - GLOBAL_RATE)
        +
        abs(test_rate - GLOBAL_RATE)
    )


    size_error = (
        abs(
            len(train_part)
            - TARGET_TRAIN
        )
        / TOTAL_ROWS
        +
        abs(
            len(val_part)
            - TARGET_VAL
        )
        / TOTAL_ROWS
        +
        abs(
            len(test)
            - TARGET_TEST
        )
        / TOTAL_ROWS
    )


    score = (
        5 * rate_error
        +
        size_error
    )


    if (
        best is None
        or score < best["score"]
    ):

        best = {
            "score": score,
            "seed": seed,

            "train":
                train_part,

            "validation":
                val_part,

            "test":
                test,

            "train_rate":
                train_rate,

            "val_rate":
                val_rate,

            "test_rate":
                test_rate,
        }


# ============================================================
# BEST SPLIT
# ============================================================

train = best["train"]
validation = best["validation"]
final_test = best["test"]


print("\n" + "=" * 90)
print("BEST CIMA TARGET-GROUP-DISJOINT SPLIT")
print("=" * 90)

print(
    "Seed:",
    best["seed"]
)

print(
    "Score:",
    round(
        best["score"],
        6
    )
)


def audit_split(
    name,
    split,
):

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(
        "Rows:",
        len(split)
    )

    print(
        "Unique target groups:",
        split[
            "target_normalized"
        ].nunique()
    )

    print(
        "Clarification positives:",
        split[
            "clarification_request"
        ].sum()
    )

    print(
        "Clarification rate:",
        round(
            split[
                "clarification_request"
            ].mean(),
            4,
        )
    )


audit_split(
    "TRAIN",
    train
)

audit_split(
    "VALIDATION",
    validation
)

audit_split(
    "FINAL TEST",
    final_test
)


# ============================================================
# GROUP LEAKAGE CHECK
# ============================================================

train_groups = set(
    train["target_normalized"]
)

val_groups = set(
    validation["target_normalized"]
)

test_groups = set(
    final_test["target_normalized"]
)


print("\n" + "=" * 90)
print("TARGET-GROUP LEAKAGE CHECK")
print("=" * 90)

print(
    "Train ∩ Validation:",
    len(
        train_groups
        & val_groups
    )
)

print(
    "Train ∩ Test:",
    len(
        train_groups
        & test_groups
    )
)

print(
    "Validation ∩ Test:",
    len(
        val_groups
        & test_groups
    )
)
```

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    confusion_matrix,
)


# ============================================================
# TARGET-ONLY MODEL
# ============================================================

target_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            ngram_range=(1, 2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
        ),
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=2000,
            random_state=42,
        ),
    ),
])


target_model.fit(
    train["target_text"],
    train["clarification_request"],
)


target_val_prob = target_model.predict_proba(
    validation["target_text"]
)[:, 1]


target_val_pred = (
    target_val_prob >= 0.50
).astype(int)


print("=" * 90)
print("CIMA MODEL A — TARGET ONLY")
print("VALIDATION SET")
print("=" * 90)


y_val = validation[
    "clarification_request"
].astype(int)


print(
    "Accuracy:",
    round(
        accuracy_score(
            y_val,
            target_val_pred
        ),
        4,
    )
)

print(
    "Balanced accuracy:",
    round(
        balanced_accuracy_score(
            y_val,
            target_val_pred
        ),
        4,
    )
)

print(
    "Precision:",
    round(
        precision_score(
            y_val,
            target_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "Recall:",
    round(
        recall_score(
            y_val,
            target_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "F1:",
    round(
        f1_score(
            y_val,
            target_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "PR-AUC:",
    round(
        average_precision_score(
            y_val,
            target_val_prob,
        ),
        4,
    )
)

print(
    "ROC-AUC:",
    round(
        roc_auc_score(
            y_val,
            target_val_prob,
        ),
        4,
    )
)

print("\nConfusion matrix:")

print(
    confusion_matrix(
        y_val,
        target_val_pred,
    )
)
```

```python
def get_recent_context(
    context,
    n_turns=4,
):

    if pd.isna(context):
        return ""

    turns = [
        x.strip()
        for x in str(context).split("\n")
        if x.strip()
    ]

    return "\n".join(
        turns[-n_turns:]
    )


for split in [
    train,
    validation,
    final_test,
]:

    split["recent_context"] = (
        split["context_text"]
        .apply(
            lambda x:
                get_recent_context(
                    x,
                    n_turns=4,
                )
        )
    )
```

```python
from sklearn.compose import ColumnTransformer


context_target_model = Pipeline([
    (
        "features",
        ColumnTransformer(
            transformers=[
                (
                    "target_tfidf",
                    TfidfVectorizer(
                        lowercase=True,
                        ngram_range=(1, 2),
                        min_df=2,
                        max_df=0.95,
                        sublinear_tf=True,
                    ),
                    "target_text",
                ),
                (
                    "context_tfidf",
                    TfidfVectorizer(
                        lowercase=True,
                        ngram_range=(1, 2),
                        min_df=2,
                        max_df=0.95,
                        sublinear_tf=True,
                    ),
                    "recent_context",
                ),
            ]
        ),
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=2000,
            random_state=42,
        ),
    ),
])


context_target_model.fit(
    train[
        [
            "target_text",
            "recent_context",
        ]
    ],
    train[
        "clarification_request"
    ],
)


context_val_prob = (
    context_target_model
    .predict_proba(
        validation[
            [
                "target_text",
                "recent_context",
            ]
        ]
    )[:, 1]
)


context_val_pred = (
    context_val_prob >= 0.50
).astype(int)


print("=" * 90)
print("CIMA MODEL B — TARGET + CONTEXT")
print("VALIDATION SET")
print("=" * 90)

print(
    "Accuracy:",
    round(
        accuracy_score(
            y_val,
            context_val_pred,
        ),
        4,
    )
)

print(
    "Balanced accuracy:",
    round(
        balanced_accuracy_score(
            y_val,
            context_val_pred,
        ),
        4,
    )
)

print(
    "Precision:",
    round(
        precision_score(
            y_val,
            context_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "Recall:",
    round(
        recall_score(
            y_val,
            context_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "F1:",
    round(
        f1_score(
            y_val,
            context_val_pred,
            zero_division=0,
        ),
        4,
    )
)

print(
    "PR-AUC:",
    round(
        average_precision_score(
            y_val,
            context_val_prob,
        ),
        4,
    )
)

print(
    "ROC-AUC:",
    round(
        roc_auc_score(
            y_val,
            context_val_prob,
        ),
        4,
    )
)

print("\nConfusion matrix:")

print(
    confusion_matrix(
        y_val,
        context_val_pred,
    )
)
```

```python
def threshold_sweep(
    model_name,
    probabilities,
    y_true,
):

    rows = []

    for threshold in np.arange(
        0.05,
        0.96,
        0.05,
    ):

        pred = (
            probabilities >= threshold
        ).astype(int)

        rows.append({
            "model":
                model_name,

            "threshold":
                round(threshold, 2),

            "precision":
                precision_score(
                    y_true,
                    pred,
                    zero_division=0,
                ),

            "recall":
                recall_score(
                    y_true,
                    pred,
                    zero_division=0,
                ),

            "f1":
                f1_score(
                    y_true,
                    pred,
                    zero_division=0,
                ),

            "balanced_accuracy":
                balanced_accuracy_score(
                    y_true,
                    pred,
                ),

            "predicted_positive":
                int(pred.sum()),
        })

    result = pd.DataFrame(rows)

    print("\n" + "=" * 90)
    print(model_name)
    print("=" * 90)

    print(
        result.to_string(
            index=False,
            float_format=lambda x: f"{x:.4f}",
        )
    )

    print("\nBest F1:")

    print(
        result.loc[
            result["f1"].idxmax()
        ]
    )

    print("\nBest balanced accuracy:")

    print(
        result.loc[
            result[
                "balanced_accuracy"
            ].idxmax()
        ]
    )

    return result


target_thresholds = threshold_sweep(
    "A_TARGET_ONLY",
    target_val_prob,
    y_val,
)


context_thresholds = threshold_sweep(
    "B_TARGET_CONTEXT",
    context_val_prob,
    y_val,
)
```
